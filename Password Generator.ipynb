```python
# Password generator

import string 
import random 

# ASk for the password length 
length_of_the_password = input(" Enter the password length: ").strip()

# Validation
# Check that the input is a number
while not length_of_the_password.isnumeric() or int(length_of_the_password) < 6: # True 'if the user enter string or any punctuations or any number less than 6' 
    if not length_of_the_password.isnumeric():
        print("Error, please enter a number")
    elif int(length_of_the_password) < 6:
        print("Error, password must be 6 or more")

    length_of_the_password = input("Enter password length: ").strip()
    
length_of_the_password = int(length_of_the_password)

# Character sets
uppercase = string.ascii_uppercase
lowercase = string.ascii_lowercase
numbers = string.digits
symbols = string.punctuation
characters = uppercase + lowercase + numbers + symbols

password = ""

for i in range(length_of_the_password):
    password +=random.choice(characters)

print(f"Your password is: {password}")
```
