```python
# Password Generator

import random
import string

name = input("Enter your name: ").strip().capitalize()

print("=" * 15)
print(f"Hello, welcome {name}")
print("=" * 15)

# User input 
password = input("Please enter the length of your password: ").strip()


# Validation to check the number is really int and not str or punc.
while not password.isnumeric() or int(password) < 6:

    # if condition here because, if the user enter a str or punc. will print this message
    if not password.isnumeric():

        print("Error, Please enter a number")

    # elif condition here because, if the user enter a number less than 6  
    elif int(password) < 6:

        print("Error, password must be 6 or more")

    # Take a access for the user to enter the length again 'if enter a number less than 6 or str or punc. will the while loop run again'
    password = input("Please enter the length of your password again: ").strip()

# Convert to int
password = int(password)


# Character sets
uppercase = string.ascii_uppercase
lowercase = string.ascii_lowercase
numbers = string.digits
symbols = string.punctuation
characters = uppercase + lowercase + numbers + symbols

# Make a empty password after the for loop will generate the password automatically depends on his user's length
thePassword = ""

for i in range(password):

    thePassword +=random.choice(characters)

print(f"Your password is: {thePassword}")
        
```
